This notebook downloads a published context, executes it locally, and prints the results. Registry access requires a CausalBench account.
```python
# Install CausalBench and the dependencies required by context 2/version 1.
# Other registry models can require different third-party libraries.
# For CUDA-specific PyTorch builds, follow https://pytorch.org/get-started/locally/ first.
%pip install --upgrade causalbench-asu "gcastle==1.0.3" torch
```

```python
# Import the public Context class.
from causalbench.modules import Context
```

```python
# Select and fetch a published context. If no valid credentials are saved,
# the request prompts for them. If this tutorial context is unavailable
# to your account, replace the ID and version with values from the catalog.
context = Context(module_id=2, version=1)

# Run every dataset-model scenario locally.
run = context.execute()
```

```python
# Print run results.
print(run)

# Publishing is an external side effect and is private by default.
# Uncomment only after reviewing the result.
# run.publish()
```
